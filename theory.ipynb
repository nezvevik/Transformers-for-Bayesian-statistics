## Stochastic Process

### Definition

Given a probability space ($\Omega, \mathcal{F}, P$), a Stochastic Process is a collection of random variables: 

$$
\{X(t) : t \in \mathcal{T}\}
$$,
where $\mathcal{T}$ is the index set and $X(t)$ is a random variable $X(t): \Omega \to \mathbb{R}$.

## Elementary Properties

#### Mean and Covariance
$$
m_X(t) = E[X(t)].
$$
$$
C_X(s, t) = \text{Cov}(X(s), X(t))
= E\!\left[(X(s) - m_X(s))(X(t) - m_X(t))\right].
$$

#### Autocorrelation Function
The autocorrelation function measures the linear dependence between $X(s)$ and $X(t)$:
$$
\rho_X(s, t) =
\frac{E[(X(s) - m_X(s))(X(t) - m_X(t))]}
{\sqrt{E[(X(s) - m_X(s))^2] \, E[(X(t) - m_X(t))^2]}}.
$$


#### Stationarity
A process is *weakly stationary* if its mean is constant and its covariance depends only on the time difference:
$$
E[X(t)] = \mu = \text{const},
\qquad
\mathrm{Cov}(X(t), X(t+\tau)) = \gamma(\tau).
$$
Here, $\gamma(\tau)$ is the *autocovariance* function that describes how are time points $t$ and $t + \tau$ correlated.

A stochastic process $ X(t) $ is said to be *strictly stationary* if the joint distribution of

$$
(X(t_1), X(t_2), \ldots, X(t_n))
$$

is identical to that of

$$
(X(t_1+ \tau), X(t_2+ \tau), \ldots, X(t_n+ \tau))
$$

for all choices of $ t_1, t_2, \ldots, t_n \in T $ and for all time shifts $ \tau $.

In other words, no matter what lag $\tau$ we chose, the process stays the same.


#### Ergodicity
A process is *ergodic* if time averages are equal to ensemble averages:
$$
\lim_{T \to \infty} \frac{1}{T} \int_0^T X(t)\, dt = E[X(t)].
$$
This means that it does not matter whether we average over time for a single realization of the process or take the average across many parallel realizations.
If the process satisfies this condition, both averages are equal.

#### Orthogonality
Procsses $X(t)$ and $Y(t)$ are *orthogonal*, when cross-correlation $\mathbb{E}[X(t) Y(t)]$ is equal to zero for all *t* and *s*.
$$
\mathbb{E}[X(t) Y(t)] = 0
$$

### **Gaussian Process**

A stochastic process $ (X(t)) $ is called a *Gaussian process* if, for any finite set of time points $ t_1, t_2, \ldots, t_n \in T $,the random vector $(X_{t_1}, X_{t_2}, \ldots, X_{t_n})$ has a **multivariate normal distribution**.

That is, there exists a mean function $ m(t) = \mathbb{E}[X(t)] $ and a covariance function $ k(s, t) = \mathrm{Cov}(X(t), X(t)) $
such that for all $ t_1, \ldots, t_n \in T $,

$$
(X_{t_1}, \ldots, X_{t_n}) \sim 
\mathcal{N} \big( 
\mathbf{m}, \mathbf{K} 
\big),
$$

where

$$
\mathbf{m} = 
\begin{bmatrix}
m_X(t_1) \\ 
\vdots \\ 
m_X(t_n)
\end{bmatrix},
\quad
\mathbf{K}_{ij} = k(t_i, t_j).
$$

and $\mathbf{K}$ is a kernel funciton.

In other words, any collection of datapints from a Gaussian process are follows a multivariate normal distribution whose mean and covariance are specified by the functions $m(t)$ and $k(s, t)$.



### Marginal Likelihood

The *marginal likelihood* is the probability of the observed data $ \mathcal{D} $ under a given model and its parameters, obtained by integrating over all possible parameter values.

Formally, for a model with parameters $ \theta $ and prior distribution $ p(\theta) $,
and likelihood $ p(\mathcal{D} \mid \theta) $, the marginal likelihood is

$$
p(\mathcal{D}) = \int p(\mathcal{D} \mid \theta) \, p(\theta) \, d\theta.
$$

It represents how well the model explains the data, and it can be used to compare different models with each other.

### Parameter estimation
We can maximize the previously defined marginal likelihood with respect to the hyperparameters. This process is called *Type II Maximum Likelihood* and for GP it has the form of this equation:

$$ \log p(\mathbf{y}\mid \theta ,x)=-{\frac {1}{2}}\left(\mathbf{y}^{\mathsf {T}}K(\theta ,x,x')^{-1}\mathbf{y}+\log \det(K(\theta ,x,x'))+n\log 2\pi \right)
$$

where $\mathbf{y}$ is the vector of target values.
